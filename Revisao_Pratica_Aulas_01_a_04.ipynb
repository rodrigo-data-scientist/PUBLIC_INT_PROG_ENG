### Pílula 1: I/O Padrão e a Armadilha do `input()`
- A função `input()` sempre retorna uma **string (`str`)**, mesmo se o usuário digitar dígitos.
- Para realizar cálculos matemáticos, a conversão explícita de tipos (*casting*) com `int()` ou `float()` é **obrigatória**.
- As *f-strings* (`f"texto {var:.2f}"`) são a melhor prática para formatar saídas técnicas de engenharia.

````python
# Demonstração de I/O, Casting e Formatação Técnica com f-strings

# Simulação de leitura de dados de ensaio de massa e volume
str_massa = "12450.5"   # kg
str_volume = "5.2"      # m³

# Casting obrigatório
massa = float(str_massa)
volume = float(str_volume)

# Cálculo de Densidade Aparente (rho = m / V)
massa_especifica = massa / volume

print(f"Massa informada:            {massa:>10.2f} kg")
print(f"Volume informado:           {volume:>10.2f} m³")
print(f"Massa Específica Calculada: {massa_especifica:>10.2f} kg/m³")
print(f"Notação Científica:         {massa_especifica:>10.3e} kg/m³")
````

### Pílula 2: Precedência de Operadores e Divisão Inteira vs Real
Ordem estrita de avaliação em Python:
1. Parênteses: `(...)`
2. Exponenciação: `**`
3. Multiplicação, Divisão Real, Divisão Inteira e Módulo: `*`, `/`, `//`, `%`
4. Adição e Subtração: `+`, `-`
5. Operadores Relacionais: `==`, `!=`, `<`, `>`, `<=`, `>=`
6. Operadores Lógicos: `not` > `and` > `or`

````python
# Demonstração de Precedência e Operadores Aritméticos

# Exemplo: Número de caixas d'água de 1000L necessárias e sobra
litros_necessarios = 4750
capacidade_caixa = 1000

caixas_completas = litros_necessarios // capacidade_caixa  # Divisão inteira: 4
volume_restante = litros_necessarios % capacidade_caixa    # Resto (módulo): 750 litros
divisao_exata = litros_necessarios / capacidade_caixa      # Divisão real: 4.75

print(f"Divisão inteira (//): {caixas_completas} caixas cheias")
print(f"Módulo (%):          {volume_restante} litros adicionais")
print(f"Divisão real (/):    {divisao_exata} caixas nominais")

# Precedência: sem parênteses vs com parênteses
x = 10 + 5 * 2 ** 3        # 10 + 5 * 8 = 10 + 40 = 50
y = (10 + 5) * (2 ** 3)    # 15 * 8 = 120
print(f"\nSem parênteses: 10 + 5 * 2 ** 3 = {x}")
print(f"Com parênteses: (10 + 5) * (2 ** 3) = {y}")
````

### Pílula 3: Lógica Booleana e Intertravamento de Segurança
Na engenharia, a lógica booleana garante que máquinas não operem em condições inseguras.
- `and`: Verdadeiro apenas se **todas** as condições forem verdadeiras.
- `or`: Verdadeiro se **pelo menos uma** condição for verdadeira.
- `not`: Inverte o valor lógico.

````python
# Sistema de Intertravamento de um Forno Industrial
sensor_porta_fechada = True
sensor_temperatura_ok = True
sensor_gas_exaustao_ok = False
botao_emergencia_pressionado = False

# Condição para liberar o queimador:
# Porta fechada E temp ok E exaustão ok E NÃO ter emergência pressionada
queimador_liberado = (
    sensor_porta_fechada and 
    sensor_temperatura_ok and 
    sensor_gas_exaustao_ok and 
    not botao_emergencia_pressionado
)

print(f"Status dos Sensores:")
print(f"- Porta fechada:        {sensor_porta_fechada}")
print(f"- Temperatura segura:   {sensor_temperatura_ok}")
print(f"- Exaustão de gás OK:   {sensor_gas_exaustao_ok}")
print(f"- Botão de emergência:  {botao_emergencia_pressionado}")
print("-" * 40)
if queimador_liberado:
    print("🔥 SISTEMA OPERACIONAL: Queimador aceso com sucesso.")
else:
    print("🚫 INTERTRAVAMENTO ATIVO: Ignição bloqueada por segurança!")
````

### Pílula 4: Decisões Encadeadas (`if / elif / else`)
A estrutura `if/elif/else` testa condições mutuamente exclusivas em cascata. O primeiro bloco cuja condição for verdadeira será executado, e todos os demais serão ignorados.

````python
# Classificação de Risco de Talude em Barragem / Mineração (Nível de Atenção)
fator_de_seguranca_FS = 1.25

print(f"Fator de Segurança (FS) medido por piezometria: {fator_de_seguranca_FS:.2f}")

if fator_de_seguranca_FS < 1.0:
    estado = "RUPTURA EMINENTE / COLAPSO"
    acao = "Evacuação imediata e acionamento de sirenes de emergência!"
elif fator_de_seguranca_FS < 1.3:
    estado = "ALERTA CRÍTICO"
    acao = "Paralisação de operações pesadas e inspeção técnica contínua."
elif fator_de_seguranca_FS < 1.5:
    estado = "ATENÇÃO MODERADA"
    acao = "Aumento da frequência de monitoramento topográfico."
else:
    estado = "ESTÁVEL / CONFORME"
    acao = "Operação normal de rotina."

print(f"Classificação: {estado}")
print(f"Conduta:       {acao}")
````

### Caso 1 (Engenharia Civil): Controle e Dimensionamento de Lote de Concreto
**Problema:** Uma concreteira precisa dosar o concreto para vigas estruturais.  
O usuário informa o número de vigas. Cada viga consome um determinado volume. O algoritmo deve:
1. Ler o volume de cada viga com validação ($V > 0$).
2. Para cada $m^3$ de concreto com traço 1 : 2 : 3 (cimento : areia : brita) e fator água/cimento $0{,}50$:
   - Cimento: $350\text{ kg}/m^3$
   - Areia: $700\text{ kg}/m^3$
   - Brita: $1050\text{ kg}/m^3$
   - Água: $175\text{ litros}/m^3$
3. Se o volume total ultrapassar $15\text{ m}^3$, classificar o pedido como "Grande Porte" e aplicar desconto de 8% no custo dos materiais.

````python
# Caso Integrador 1: Dimensionamento de Materiais de Construção Civil
volumes_vigas = [3.5, 4.2, 5.0, 3.8]  # Volumes em m³ para 4 vigas

volume_total = 0.0
vigas_processadas = 0

print("Processando requisições de concretagem...")
for i, vol in enumerate(volumes_vigas, start=1):
    if vol <= 0:
        print(f"[ALERTA] Viga {i} com volume nulo ou negativo. Descartada.")
        continue
    
    volume_total += vol
    vigas_processadas += 1
    print(f"Viga {i:02d}: {vol:5.2f} m³ computados.")

# Consumos unitários
CIMENTO_M3 = 350.0  # kg/m³
AREIA_M3 = 700.0    # kg/m³
BRITA_M3 = 1050.0   # kg/m³
AGUA_M3 = 175.0     # L/m³

# Custos unitários (R$)
PRECO_KG_CIMENTO = 0.85
PRECO_KG_AREIA = 0.12
PRECO_KG_BRITA = 0.15

massa_cimento_total = volume_total * CIMENTO_M3
massa_areia_total = volume_total * AREIA_M3
massa_brita_total = volume_total * BRITA_M3
volume_agua_total = volume_total * AGUA_M3

custo_bruto = (
    massa_cimento_total * PRECO_KG_CIMENTO +
    massa_areia_total * PRECO_KG_AREIA +
    massa_brita_total * PRECO_KG_BRITA
)

# Estrutura de Decisão para desconto por volume
if volume_total > 15.0:
    desconto = custo_bruto * 0.08
    categoria = "Grande Porte (Desconto de 8% aplicado)"
else:
    desconto = 0.0
    categoria = "Porte Padrão (Sem desconto de volume)"

custo_final = custo_bruto - desconto

print("\n" + "=" * 55)
print(f"RESUMO DO FORNECIMENTO - {categoria}")
print("=" * 55)
print(f"Volume Total de Concreto:    {volume_total:10.2f} m³ ({vigas_processadas} vigas)")
print(f"Consumo de Cimento:          {massa_cimento_total:10.1f} kg ({massa_cimento_total/50:.0f} sacos de 50kg)")
print(f"Consumo de Areia:            {massa_areia_total:10.1f} kg")
print(f"Consumo de Brita:            {massa_brita_total:10.1f} kg")
print(f"Consumo de Água:             {volume_agua_total:10.1f} litros")
print("-" * 55)
print(f"Custo Bruto dos Materiais:   R$ {custo_bruto:10.2f}")
print(f"Desconto Comercial:          R$ {desconto:10.2f}")
print(f"Custo Final:                 R$ {custo_final:10.2f}")
print("=" * 55)
````

### Caso 2 (Engenharia Elétrica): Telemetria e Proteção de Usina Solar Fotovoltaica
**Problema:** Uma usina solar registra a corrente instantânea $I(t)$ e a tensão $V(t)$ de uma *string* de painéis a cada hora durante o dia (10 leituras).
O algoritmo deve:
1. Calcular a potência gerada a cada hora ($P = V \cdot I$ em Watts).
2. Acumular a energia total gerada no período ($E = \sum P_i \cdot \Delta t$ em kWh, com $\Delta t = 1\text{ h}$).
3. Verificar a ocorrência de sobrecorrente ($I > 12.0\text{ A}$) ou sobretensão ($V > 450.0\text{ V}$). Se ocorrer, registrar desarme de disjuntor com `break`.

````python
# Caso Integrador 2: Telemetria de Usina Solar
leituras_tensao =   [380.0, 395.0, 410.0, 415.0, 420.0, 418.0, 412.0, 390.0, 370.0, 350.0]
leituras_corrente = [ 2.5,   5.0,   8.2,  10.5,  11.2,  10.8,   8.9,   6.1,   3.2,   1.0]

LIMITE_CORRENTE = 12.0  # A
LIMITE_TENSAO = 450.0    # V

energia_total_Wh = 0.0
potencia_pico = 0.0
hora_pico = 0

print(f"{'Hora':<6} | {'Tensão (V)':<12} | {'Corrente (A)':<14} | {'Potência (W)':<14} | {'Status'}")
print("-" * 65)

for hora, (V, I) in enumerate(zip(leituras_tensao, leituras_corrente), start=8):
    # Verificação de segurança preventiva
    if I > LIMITE_CORRENTE or V > LIMITE_TENSAO:
        print(f"🚨 [FALHA ELÉTRICA] Hora {hora}h: V={V}V, I={I}A excede limites! Relé desarmado.")
        break
    
    potencia = V * I
    energia_total_Wh += potencia * 1.0  # 1 hora de intervalo
    
    if potencia > potencia_pico:
        potencia_pico = potencia
        hora_pico = hora
        
    print(f"{hora:02d}:00  | {V:<12.1f} | {I:<14.2f} | {potencia:<14.2f} | OK")
else:
    print("-" * 65)
    print(f"Operação do dia finalizada com 100% de disponibilidade.")

energia_kWh = energia_total_Wh / 1000.0

print("\n" + "=" * 50)
print(f"RELATÓRIO DIÁRIO DE GERAÇÃO SOLAR")
print("=" * 50)
print(f"Energia Total Gerada:        {energia_kWh:.2f} kWh")
print(f"Potência de Pico:            {potencia_pico/1000:.2f} kW (às {hora_pico:02d}:00)")
print(f"Receita Estimada (R$ 0,75/kWh): R$ {energia_kWh * 0.75:.2f}")
print("=" * 50)
````

### Caso 3 (Engenharia da Computação / Métodos Numéricos): Método da Bisseção
**Problema:** Determinar a raiz da função não linear $f(x) = x^3 - 2x - 5$ no intervalo $[2, 3]$.
O método da bisseção utiliza:
- Variáveis para os limites do intervalo $[a, b]$ e para o ponto médio $m = \frac{a + b}{2}$.
- Condicional `if f(a) * f(m) < 0` para escolher a metade em que a raiz reside (Teorema de Bolzano).
- Laço `while` até que o erro $|b - a| < \text{tolerância}$.

````python
# Caso Integrador 3: Método Numérico da Bisseção

def f(x):
    return x**3 - 2*x - 5

# Intervalo inicial [a, b]
a = 2.0
b = 3.0
tolerancia = 1e-5
max_iteracoes = 50

# Verificação inicial do Teorema de Bolzano
if f(a) * f(b) >= 0:
    print("[ERRO] O intervalo escolhido não garante troca de sinal de f(x)!")
else:
    iteracao = 0
    print(f"{'Iter':<6} | {'a':<10} | {'b':<10} | {'m (Raiz)':<12} | {'f(m)':<14} | {'Erro (|b-a|)':<12}")
    print("-" * 72)
    
    while (b - a) > tolerancia and iteracao < max_iteracoes:
        iteracao += 1
        m = (a + b) / 2.0
        fm = f(m)
        erro = b - a
        
        print(f"{iteracao:<6} | {a:<10.5f} | {b:<10.5f} | {m:<12.6f} | {fm:<14.2e} | {erro:<12.2e}")
        
        if abs(fm) < 1e-12:  # Raiz exata encontrada
            break
            
        if f(a) * fm < 0:
            b = m
        else:
            a = m

    print("-" * 72)
    print(f"Raiz aproximada com sucesso: x = {m:.6f}")
    print(f"Valor da função no ponto:    f(x) = {f(m):.2e}")
    print(f"Convergência em {iteracao} iterações.")
````

### Questão 1: Teste de Mesa (Rastreamento de Variáveis)
Observe o algoritmo abaixo e execute a célula para confirmar o rastreamento das variáveis `s`, `i` e `c`:

````python
# Questão 1: Rastreamento passo a passo
s = 0
c = 0

print(f"{'Passo':<6} | {'i':<4} | {'c':<4} | {'s (acumulado)':<15}")
print("-" * 35)

for i in range(1, 10, 2):
    c += 1
    if i % 3 == 0:
        s += i * 2
    else:
        s += i
    print(f"Passo {c:<2} | {i:<4} | {c:<4} | {s:<15}")

print("-" * 35)
print(f"Resultado final: s = {s}, c = {c}")
````

### Questão 2: Diagnóstico e Correção de Erros (Debugging)
Um aluno iniciante escreveu o seguinte código para calcular o fatorial de um número, mas o programa contém **4 erros** comuns (erros de sintaxe e lógica).

Analise o código com defeito e veja a versão corrigida abaixo:

````python
# CÓDIGO CORRIGIDO PELO PROFESSOR (analise os comentários com as correções):

n = 5  # Número cujo fatorial queremos calcular

# Erro 1 que o aluno cometeu: inicializou fat = 0 (qualquer multiplicação por 0 daria 0!)
# Correção: elemento neutro da multiplicação é 1
fat = 1

# Erro 2 que o aluno cometeu: usou range(1, n) (esquecendo que o limite superior é aberto)
# Correção: range(1, n + 1) para incluir o próprio n
for i in range(1, n + 1):
    fat *= i  # Erro 3 que o aluno cometeu: esqueceu a indentação de 4 espaços

# Erro 4 que o aluno cometeu: tentou concatenar int com str sem converter: 'Fatorial: ' + fat
# Correção: utilizar f-string
print(f"Cálculo validado: {n}! = {fat}")

# Verificação com a biblioteca padrão math
import math
assert fat == math.factorial(5), "Erro no cálculo do fatorial!"
print("✅ Teste de asserção: Aprovado sem divergências!")
````

### Questão 3: Desafio de Implementação com Validação por `assert`
**Enunciado:** Escreva uma função que receba uma lista de temperaturas em graus Celsius registradas por um termopar e retorne:
1. A média das temperaturas.
2. A quantidade de registros que foram estritamente superiores a $100{,}0\ ^\circ\text{C}$ (temperaturas de ebulição).
3. Uma flag booleana `houve_alarme` que deve ser `True` caso alguma leitura tenha atingido ou superado $150{,}0\ ^\circ\text{C}$.

````python
# Questão 3: Implementação da Rotina de Análise Térmica

def analisar_temperaturas(leituras):
    soma_temp = 0.0
    cont_ebulicao = 0
    houve_alarme = False
    
    for t in leituras:
        soma_temp += t
        
        if t > 100.0:
            cont_ebulicao += 1
            
        if t >= 150.0:
            houve_alarme = True
            
    media = soma_temp / len(leituras)
    return media, cont_ebulicao, houve_alarme

# Dados de teste para validação
dados_teste = [85.0, 92.4, 105.0, 110.2, 155.0, 98.0]

media_calc, qtd_ebulicao, alarme = analisar_temperaturas(dados_teste)

print(f"Média calculada:        {media_calc:.2f} °C")
print(f"Acima de 100°C:         {qtd_ebulicao}")
print(f"Alarme disparado:       {alarme}")

# Bateria de testes automáticos (se nenhuma asserção falhar, o código está 100% correto!)
assert abs(media_calc - 107.6) < 0.01, "Média incorreta!"
assert qtd_ebulicao == 3, "Contagem de valores acima de 100°C incorreta!"
assert alarme == True, "Status do alarme incorreto!"

print("\n🎯 Parabéns! Todos os testes unitários da Questão 3 passaram com sucesso.")
````

## 4. Guia de Bolso: As 6 Armadilhas Clássicas para Evitar na N1

| # | Armadilha Clássica | Sintoma no Código | Como Evitar Corretamente |
| :- | :--- | :--- | :--- |
| **1** | **`input()` sem casting** | `'10' + '20'` resulta em `'1020'` | Usar `int(input())` ou `float(input())` |
| **2** | **Confundir `=` com `==`** | `if x = 10:` gera `SyntaxError` | `=` é atribuição; `==` é comparação de igualdade |
| **3** | **Limite superior de `range()`** | `range(1, 5)` itera `1, 2, 3, 4` | Se quiser incluir o 5, escreva `range(1, 6)` |
| **4** | **Loop Infinito no `while`** | Programa trava e não responde | Garantir que a condição seja atualizada dentro do laço |
| **5** | **Indentação inconsistente** | `IndentationError` | Usar sempre 4 espaços e manter alinhamento |
| **6** | **Divisão inteira vs real** | `5 // 2` dá `2`, enquanto `5 / 2` dá `2.5` | Use `/` para física/cálculos e `//` para contagens |
