## 1. O Paralelo Multi-Linguagem: Repetições na Engenharia

Antes de mergulhar na sintaxe do Python, vejamos como as diferentes linguagens tratam repetições:

| Conceito / Linguagem | Portugol (VisualG/Studio) | Python (Linguagem Principal) | C++ (Alto Desempenho) | Rust (Segurança de Memória) |
| :--- | :--- | :--- | :--- | :--- |
| **Laço Contado** | `para i de 1 ate N faca` | `for i in range(1, N + 1):` | `for (int i = 1; i <= N; ++i)` | `for i in 1..=N` |
| **Laço Condicional** | `enquanto (cond) faca` | `while cond:` | `while (cond)` | `while cond` |
| **Laço Infinito Explícito** | `enquanto (verdadeiro) faca` | `while True:` | `while (true)` ou `for(;;)` | `loop { ... }` |
| **Interrupção Prematura** | `interrompa` | `break` | `break;` | `break;` |
| **Próxima Iteração** | Não padronizado | `continue` | `continue;` | `continue;` |

> 💡 **Nota na Engenharia:** Em Python, o laço `for` é essencialmente um iterador sobre sequências ou coleções. Em C++ e Rust, loops contados são compilados para instruções assembly de salto ultrarrápidas, ideais para controladores de motores e processamento de sinais em tempo real.

## 2. Laços Contados: A Estrutura `for` e a Função `range()`

O laço `for` é utilizado quando sabemos previamente (ou quando podemos determinar em tempo de execução) quantas iterações devem ocorrer.

A função `range()` gera uma progressão aritmética de números inteiros:
- `range(stop)`: vai de `0` até `stop - 1` com passo `1`.
- `range(start, stop)`: vai de `start` até `stop - 1` com passo `1`.
- `range(start, stop, step)`: vai de `start` até `stop - 1` variando de `step` em `step`.

Execute as células a seguir para observar as variações:

````python
# Demonstração das três formas da função range()

print("--- Forma 1: range(5) -> de 0 até 4 ---")
for i in range(5):
    print(f"Iteração: {i}")

print("\n--- Forma 2: range(1, 6) -> de 1 até 5 ---")
for i in range(1, 6):
    print(f"Contagem: {i}")

print("\n--- Forma 3: range(0, 20, 5) -> passo positivo 5 ---")
for valor in range(0, 21, 5):
    print(f"Nível de pressão: {valor} kPa")

print("\n--- Forma 4: range(10, 0, -2) -> contagem regressiva ---")
for cont in range(10, 0, -2):
    print(f"Tempo restante para acionamento: {cont} s")
````

## 3. Padrões Algorítmicos: Contador e Acumulador

Em engenharia, frequentemente precisamos:
1. **Contador:** Variável inteira que incrementa a cada evento (ex.: contar quantas barras de aço excederam a tensão admissível).
2. **Acumulador / Somatório:** Variável que acumula valores numéricos sucessivos (ex.: somatório de forças, cálculo de massa total, energia consumida).

````python
# Padrão Acumulador e Contador: Cálculo do carregamento total em uma laje
# Suponha 6 vigas com diferentes cargas pontuais distribuídas (em kN)

cargas_vigas = [12.5, 18.0, 9.4, 22.1, 15.0, 8.8]

carga_total = 0.0      # Variável Acumuladora
vigas_criticas = 0     # Variável Contadora (cargas acima de 15 kN)

for i in range(len(cargas_vigas)):
    carga = cargas_vigas[i]
    carga_total += carga  # Acumulação
    
    if carga > 15.0:
        vigas_criticas += 1
        print(f"[ALERTA] Viga {i+1}: Carga de {carga:.1f} kN excede o limite operacional!")

media_carga = carga_total / len(cargas_vigas)

print("\n" + "="*45)
print(f"Carga total atuante na estrutura: {carga_total:.2f} kN")
print(f"Carga média por viga:             {media_carga:.2f} kN")
print(f"Quantidade de vigas críticas:    {vigas_criticas} de {len(cargas_vigas)}")
print("="*45)
````

## 4. Aplicação de Engenharia: Centroide de Seção Transversal Composta

Na Engenharia Civil e Mecânica, a posição do centroide vertical ($ar{y}$) de uma seção composta por $N$ figuras geométricas é dada por:

$$\bar{y} = \frac{\sum_{i=1}^{N} A_i \cdot y_i}{\sum_{i=1}^{N} A_i}$$

Onde:
- $A_i$ é a área de cada elemento discretizado ($mm^2$).
- $y_i$ é a coordenada vertical do centroide de cada elemento em relação à base ($mm$).

````python
# Cálculo computacional do Centroide usando laço for e acumuladores
# Perfil 'T' discretizado em mesa (flange) e alma (web)

areas = [600.0, 1000.0]       # Área em mm² (mesa: 60x10, alma: 10x100)
alturas_y = [105.0, 50.0]     # Coordenada y_i do centroide de cada parte em mm

somatorio_numerador = 0.0     # Acumulador do momento estático de 1ª ordem (A_i * y_i)
somatorio_denominador = 0.0   # Acumulador da área total (A_i)

n_elementos = len(areas)

print(f"{'Elemento':<10} | {'Área (mm²)':<12} | {'y_i (mm)':<10} | {'A_i * y_i (mm³)':<15}")
print("-" * 55)

for i in range(n_elementos):
    a = areas[i]
    y = alturas_y[i]
    momento_estatico = a * y
    
    somatorio_numerador += momento_estatico
    somatorio_denominador += a
    
    print(f"Elem {i+1:<5} | {a:<12.1f} | {y:<10.1f} | {momento_estatico:<15.1f}")

y_barra = somatorio_numerador / somatorio_denominador

print("-" * 55)
print(f"Área Total da Seção:       {somatorio_denominador:.1f} mm²")
print(f"Momento Estático Total:    {somatorio_numerador:.1f} mm³")
print(f"Posição do Centroide (ȳ):  {y_barra:.2f} mm em relação à base")
````

## 5. Aplicação de Engenharia: Tabela de Queda de Tensão em Condutor Elétrico

Na Engenharia Elétrica, a resistência de um condutor de cobre com seção $S$ ($mm^2$) e comprimento $L$ ($m$) é:

$$R = \rho \cdot \frac{L}{S}$$

Onde a resistividade do cobre é $\rho = 0{,}0172\ \Omega \cdot mm^2 / m$.  
A queda de tensão para um circuito monofásico a dois condutores conduzindo corrente $I$ é:

$$\Delta V = 2 \cdot R \cdot I = 2 \cdot \left(\rho \cdot \frac{L}{S}\right) \cdot I$$

Vamos gerar a tabela técnica para comprimentos de 10 a 100 metros (passo de 10 metros):

````python
# Parâmetros elétricos do circuito
rho_cobre = 0.0172    # Ohm * mm² / m
secao_cabo = 4.0      # mm²
corrente_I = 25.0     # Ampères (A)
tensao_nominal = 220.0 # Volts (V)

print(f"Simulação para Seção = {secao_cabo} mm² | Corrente = {corrente_I} A | Tensão = {tensao_nominal} V")
print("=" * 65)
print(f"{'L (m)':<8} | {'Resistência (Ω)':<18} | {'Queda ΔV (V)':<14} | {'Queda (%)':<10} | {'Status'}")
print("-" * 65)

# Laço for com passo de 10 metros
for L in range(10, 101, 10):
    R = rho_cobre * (L / secao_cabo)
    delta_V = 2.0 * R * corrente_I
    queda_percentual = (delta_V / tensao_nominal) * 100.0
    
    # NBR 5410: Queda de tensão máxima admissível em circuitos terminais é 4%
    status = "OK (Conforme)" if queda_percentual <= 4.0 else "NÃO CONFORME (>4%)"
    
    print(f"{L:<8} | {R:<18.4f} | {delta_V:<14.2f} | {queda_percentual:<9.2f}% | {status}")

print("=" * 65)
````

## 6. Laços Condicionais: A Estrutura `while`

O laço `while` repete um bloco de código **enquanto** uma condição booleana for verdadeira. É ideal quando **não sabemos de antemão** o número exato de repetições (ex.: convergência numérica, esvaziamento de tanque, leitura de sensor até botão ser pressionado).

### Anatomia do `while`:
```python
inicializacao()
while condicao_de_guarda:
    # acoes executadas repetidamente
    atualizacao_da_condicao() # OBRIGATÓRIO para evitar loop infinito!
```

> ⚠️ **Atenção:** Se a condição nunca se tornar `False`, o programa entrará em **loop infinito**. Em tempo de execução, interrompa com o botão **Stop (Interrupt Kernel)** no Jupyter.

````python
# Exemplo de laço while: Esvaziamento de um tanque industrial de efluentes
# Altura inicial da coluna de água h = 5.0 m.
# A cada minuto, a vazão de saída reduz a lâmina de água segundo h(t) = h - 0.15 * sqrt(h)
# O processo cessa quando h <= 0.20 m (resíduo de fundo)

import math

h = 5.0             # Altura inicial em metros
minuto = 0          # Contador de tempo
taxa_vazao = 0.35   # Coeficiente de descarga da válvula

print(f"{'Tempo (min)':<12} | {'Nível do Tanque h (m)':<25} | {'Estado'}")
print("-" * 50)

while h > 0.20:
    minuto += 1
    variacao = taxa_vazao * math.sqrt(h)
    h -= variacao
    if h < 0.20:
        h = 0.20  # Nível mínimo de proteção
    
    print(f"{minuto:<12} | {h:<25.3f} | Operando drenagem")

print("-" * 50)
print(f"Processo finalizado em {minuto} minutos. Nível final residual: {h:.2f} m.")
````

## 7. Padrão de Validação Robusta de Entrada

Em sistemas de controle de engenharia, nunca devemos aceitar valores fisicamente impossíveis (ex.: temperaturas abaixo de 0 Kelvin, resistências negativas, diâmetros nulos). O laço `while` é a ferramenta perfeita para forçar a digitação correta.

````python
# Validação de Entrada simulada
# Demonstração com lista de tentativas para execução automática em sala

entradas_teste = [-15.0, 0.0, 1500.0, 85.0]  # Usuário digitou valores inválidos até digitar 85.0
indice_leitura = 0

print("Iniciando rotina de configuração da pressão de teste hidrostático da tubulação...")

pressao = entradas_teste[indice_leitura]
indice_leitura += 1

# Pressão aceitável para o ensaio: entre 10.0 e 200.0 bar
while pressao < 10.0 or pressao > 200.0:
    print(f"[ERRO DE SEGURANÇA] Valor lido: {pressao:.1f} bar é INVÁLIDO! (Faixa permitida: 10 a 200 bar)")
    # Simula próxima leitura do operador
    pressao = entradas_teste[indice_leitura]
    indice_leitura += 1

print(f"\n[SUCESSO] Pressão calibrada com segurança: {pressao:.1f} bar. Liberando bombeamento.")
````

## 8. Modificadores de Fluxo: `break`, `continue` e a Cláusula `else`

- `break`: Interrompe imediatamente o laço e pula para a primeira linha fora dele (muito usado em disparos de emergência / falha crítica).
- `continue`: Interrompe apenas a iteração atual e salta diretamente para a verificação da próxima (muito usado para descartar ruído em sensores).
- `else` no laço: Bloco executado quando o laço termina naturalmente **sem ter atingido nenhum `break`**.

````python
# Monitoramento térmico de um transformador de potência
# Leituras de temperatura (°C) registradas a cada segundo

leituras_sensores = [42.1, 45.0, -999.0, 50.3, 58.7, 72.4, 91.2, 60.0]
TEMPERATURA_CRITICA = 85.0  # Limite de desarme por relé térmico

print("Iniciando varredura de telemetria térmica...")

for segundo, temp in enumerate(leituras_sensores, start=1):
    # Caso 1: Leitura de erro de sensor (-999.0 indica cabo desconectado)
    if temp < -50.0:
        print(f"Segundo {segundo}: Leitura de ruído ({temp}°C) ignorada com 'continue'.")
        continue
    
    print(f"Segundo {segundo}: Temperatura = {temp:.1f} °C")
    
    # Caso 2: Superaquecimento crítico (desarme imediato)
    if temp >= TEMPERATURA_CRITICA:
        print(f"\n🚨 [DESARME DE EMERGÊNCIA - BREAK] No segundo {segundo}, temperatura de {temp:.1f}°C excedeu {TEMPERATURA_CRITICA}°C!")
        print("Disjuntor principal aberto para salvar o transformador.")
        break
else:
    # Só executa se NENHUM break ocorreu
    print("\n✅ Ciclo de ensaio concluído com sucesso. Nenhuma anomalia de temperatura detectada.")
````

## 9. Métodos Numéricos e Convergência: Método de Newton-Raphson

Grande parte dos problemas práticos de engenharia não possui solução analítica fechada, exigindo métodos iterativos.
Para calcular a raiz quadrada de um número real $S > 0$ ($x = \sqrt{S}$), a relação de recorrência de Newton-Raphson (também chamada de Método Babilônico) é:

$$x_{k+1} = \frac{1}{2} \left( x_k + \frac{S}{x_k} \right)$$

O laço continua até que o erro absoluto $|x_{k+1} - x_k| < \text{tolerância}$ ou o limite de iterações de segurança seja atingido:

````python
# Implementação do Método de Newton-Raphson para raiz quadrada
S = 50.0            # Valor do qual queremos a raiz quadrada (esperado: ~7.0710678)
tolerancia = 1e-7   # Erro máximo aceitável (10^-7)
max_iter = 100      # Trava de segurança contra loop infinito

x_atual = S / 2.0   # Estimativa inicial
iteracao = 0
erro = 1.0

print(f"{'Iteração':<10} | {'Estimativa x_k':<20} | {'Erro (|x_k+1 - x_k|)':<22}")
print("-" * 60)

while erro > tolerancia and iteracao < max_iter:
    iteracao += 1
    x_proximo = 0.5 * (x_atual + S / x_atual)
    erro = abs(x_proximo - x_atual)
    
    print(f"{iteracao:<10} | {x_atual:<20.9f} | {erro:<22.9e}")
    x_atual = x_proximo

print("-" * 60)
print(f"Raiz calculada:  {x_atual:.9f}")
print(f"Raiz via math:   {math.sqrt(S):.9f}")
print(f"Convergência alcançada em {iteracao} iterações!")
````

## 10. Laços Aninhados: Varredura de Malha 2D de Engenharia

Quando discretizamos uma placa metálica, uma laje ou um terreno em uma grade bidimensional de coordenadas $(x, y)$, utilizamos **laços aninhados** (um laço dentro de outro).

Vamos simular a estimativa da temperatura em uma chapa condutora quadrada com 5 nós em $X$ e 5 nós em $Y$, dada pela função analítica simplificada:
$$T(x, y) = 20 + 2 \cdot x^2 + 3 \cdot y$$

````python
# Malha 2D: Laços Aninhados
n_pontos_x = 5
n_pontos_y = 5

print("Distribuição de Temperatura na Placa Metálica (°C):")
print(f"{'y\\x':<6}", end="")
for x in range(n_pontos_x):
    print(f"x={x:<6}", end="")
print("\n" + "-" * 42)

for y in range(n_pontos_y):
    print(f"y={y:<4} |", end="")
    for x in range(n_pontos_x):
        temp = 20.0 + 2.0 * (x ** 2) + 3.0 * y
        print(f"{temp:6.1f} ", end="")
    print()  # Quebra de linha ao final de cada linha da malha
````

### Desafio 1: Série de Taylor para o Cosseno
A função cosseno pode ser aproximada pela série de Taylor:
$$\cos(x) = \sum_{n=0}^{\infty} (-1)^n \frac{x^{2n}}{(2n)!} = 1 - \frac{x^2}{2!} + \frac{x^4}{4!} - \frac{x^6}{6!} + \dots$$

Escreva um algoritmo com laço `while` que calcule a aproximação de $\cos(x)$ até que o valor absoluto do termo somado seja menor que $\epsilon = 10^{-6}$.

````python
# Resolução do Desafio 1: Série de Taylor
import math

angulo_graus = 45.0
x = math.radians(angulo_graus) # Conversão para radianos

cos_aprox = 0.0
n = 0
termo = 1.0
tolerancia = 1e-6

while abs(termo) > tolerancia:
    # Termo geral da série de Taylor do cosseno
    termo = ((-1) ** n) * (x ** (2 * n)) / math.factorial(2 * n)
    cos_aprox += termo
    n += 1

print(f"Ângulo:                 {angulo_graus}° ({x:.4f} rad)")
print(f"Cosseno aproximado:     {cos_aprox:.8f} (em {n} termos)")
print(f"Cosseno via math.cos:   {math.cos(x):.8f}")
print(f"Diferença residual:     {abs(cos_aprox - math.cos(x)):.2e}")
````

### Desafio 2: Ensaio Tecnológico de Corpos de Prova de Concreto (NBR 5739)
Uma equipe de Engenharia Civil realizou o ensaio de ruptura à compressão em 8 corpos de prova cilíndricos ($f_{ci}$ em MPa).
O projeto exige $f_{ck} = 30.0\text{ MPa}$.
Com auxílio de um laço `for`:
1. Identifique o corpo de prova com maior resistência e o com menor resistência.
2. Calcule a resistência média da amostragem.
3. Conte quantos corpos de prova ficaram abaixo da especificação ($< 30\text{ MPa}$).

````python
# Resolução do Desafio 2: Análise de CPs de Concreto
resistencias = [31.5, 29.2, 33.8, 30.1, 28.5, 34.0, 32.2, 30.8]

soma_resistencias = 0.0
maior_f = resistencias[0]
menor_f = resistencias[0]
abaixo_fck = 0
fck_projeto = 30.0

for r in resistencias:
    soma_resistencias += r
    
    if r > maior_f:
        maior_f = r
        
    if r < menor_f:
        menor_f = r
        
    if r < fck_projeto:
        abaixo_fck += 1

media_f = soma_resistencias / len(resistencias)

print("="*45)
print("RELATÓRIO DE CONTROLE TECNOLÓGICO DE CONCRETO")
print("="*45)
print(f"Total de CPs ensaiados:          {len(resistencias)}")
print(f"Resistência Média:               {media_f:.2f} MPa")
print(f"Maior Resistência observada:     {maior_f:.2f} MPa")
print(f"Menor Resistência observada:     {menor_f:.2f} MPa")
print(f"CPs Não Conformes (< {fck_projeto} MPa):   {abaixo_fck} ({abaixo_fck/len(resistencias)*100:.1f}%)")
print("="*45)
````
